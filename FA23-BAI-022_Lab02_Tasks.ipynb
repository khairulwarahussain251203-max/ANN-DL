### Task 3: Addition & Subtraction

```python
term1 = np.array([
    [78, 82, 90],
    [65, 70, 60],
    [88, 95, 91]
])

term2 = np.array([
    [85, 80, 92],
    [70, 75, 68],
    [90, 93, 96]
])

print("Term 1:\n", term1)
print("Term 2:\n", term2)
```

```python
combined_marks = term1 + term2
print("Combined marks (Term1 + Term2):")
print(combined_marks)
```

```python
improvement = term2 - term1
print("Improvement (Term2 - Term1):")
print(improvement)
```

```python
total_improvement_per_student = improvement.sum(axis=1)
best_student_index = np.argmax(total_improvement_per_student)

print("Total improvement per student:", total_improvement_per_student)
print("Student with highest total improvement: Student", best_student_index + 1)
```

### Task 4: Dot Product

```python
item_prices = np.array([50, 20, 15, 100])
quantities = np.array([2, 5, 3, 1])

total_bill = np.dot(item_prices, quantities)
print("Total bill:", total_bill)
```

```python
customer_quantities = np.array([
    [2, 5, 3, 1],
    [1, 2, 0, 4]
])

# For the dot product to work, the number of columns in customer_quantities (4)
# must match the number of elements in item_prices (4), since each item price
# is multiplied with the matching quantity and summed up.
total_bills = np.dot(customer_quantities, item_prices)
print("Total bill for each customer:", total_bills)
```

### Task 8: Random Matrices

```python
seed_value = 42

np.random.seed(seed_value)
decimals = np.random.rand(3, 3)
print("Random decimals (0-1):")
print(decimals)
```

```python
random_ints = np.random.randint(10, 101, size=(3, 3))
print("Random integers (10-100):")
print(random_ints)
```

```python
standard_normal = np.random.randn(4, 4)
print("Standard normal 4x4:")
print(standard_normal)
```

```python
np.random.seed(seed_value)
run1 = np.random.rand(3, 3)

np.random.seed(seed_value)
run2 = np.random.rand(3, 3)

print("Run 1:\n", run1)
print("Run 2:\n", run2)
print("Identical?", np.array_equal(run1, run2))
# Both outputs are identical because resetting the same seed before each call
# makes NumPy's pseudo-random generator start from the exact same internal state
```

### Task 9: Transpose

```python
performance = np.array([
    [80, 85, 88, 90, 92],
    [70, 75, 78, 74, 80],
    [95, 96, 94, 98, 97]
])
print("Original shape:", performance.shape)
```

```python
performance_T = performance.T
print("Transposed shape:", performance_T.shape)
print(performance_T)
```

```python
double_transpose = performance.T.T
print("Transposing twice gives back original?", np.array_equal(performance, double_transpose))
```

```python
matrix_a = np.array([
    [1, 2],
    [3, 4],
    [5, 6],
    [7, 8]
])

matrix_b = np.array([
    [1, 0],
    [0, 1],
    [1, 1]
])

result = np.dot(matrix_a, matrix_b.T)
print("Result of dot product (4x2) . (3x2).T:")
print(result)
print("Result shape:", result.shape)
```
